# ANN for Regression - boston housing dataset

### 1.Import necessary Libraries

```python
import pandas as pd
import keras
from keras.models import Sequential
from keras.layers import Dense
from sklearn.preprocessing import StandardScaler
import numpy as np
```

### 2.Import datasets

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.boston_housing.load_data()
```

### 3.Data understandng

```python
x_train.shape,y_train.shape
```

```python
x_test.shape,y_test.shape
```

### 4.Data Preparation - normalization

```python
scalar = StandardScaler()
x_train=scalar.fit_transform(x_train)
x_test=scalar.fit_transform(x_test)
```

```python
x_train.std().round(),x_train.mean().round()
```

### 5.Model building

```python
model = Sequential()
model.add(Dense( units=64,activation="relu",input_shape=(13,)))
model.add(Dense( units=32,activation="relu"))
model.add(Dense( units=16,activation="relu"))
model = Sequential()
model.add(Dense(1))
```

```python
model.compile( optimizer='adam',loss='mse',metrics=['mae'])
```

### 6.model training | 7.model testing | 8.model evaluation

```python
model.fit(x=x_train, y=y_train,batch_size=16, epochs=100,validation_split=0.2, verbose=2)
```

### 9.Model Prediction

```python
new_data = np.array([[0.02, 18, 2.5, 0, 0.45, 6.5, 65, 4.5, 1, 300, 15, 390, 5]])
new_data = scalar.transform(new_data)

# Predict
prediction = model.predict(new_data)

print("Predicted House Price:", prediction)
```
